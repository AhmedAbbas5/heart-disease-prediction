## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import pickle

from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report
)

print("Libraries imported successfully.")
```

## 2. Download and Load the UCI Cleveland Heart Disease Dataset

```python
# UCI processed Cleveland dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/heart-disease/processed.cleveland.data"

columns = [
    "age", "sex", "cp", "trestbps", "chol", "fbs", "restecg",
    "thalach", "exang", "oldpeak", "slope", "ca", "thal", "target"
]

df = pd.read_csv(url, header=None, names=columns, na_values="?")

print("Dataset shape:", df.shape)
df.head()
```

## 3. Understand the Data

```python
print(df.info())
print("\nMissing values:")
print(df.isnull().sum())
```

```python
print("Target values before conversion:")
print(df["target"].value_counts(dropna=False))
```

## 4. Preprocessing

The assignment restricts the model to **exactly four input attributes**.  
The selected attributes are `age`, `sex`, `cp`, and `thalach`.

The UCI target is 0–4. We convert all values greater than 0 to 1 so the task becomes binary classification.


```python
# Convert target to binary: 0 = no disease, 1-4 = disease
df["target"] = (df["target"] > 0).astype(int)

# Keep exactly four input attributes
features = ["age", "sex", "cp", "thalach"]

X = df[features].copy()
y = df["target"].copy()

# Handle missing values in the selected inputs, if any
X = X.fillna(X.median())

print("Input features:", list(X.columns))
print("Number of input features:", X.shape[1])
print("Target classes:", sorted(y.unique()))
```

## 5. Train/Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 6. Train the Support Vector Classifier

```python
model = SVC()

model.fit(X_train, y_train)

print("Model training completed.")
```

## 7. Evaluate the Model

```python
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)

print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

## 8. Save the Trained Model

```python
with open("heart_model.pkl", "wb") as file:
    pickle.dump(model, file)

print("Saved: heart_model.pkl")
```

## 9. Test One Prediction

```python
# Example patient: age, sex, chest-pain type, maximum heart rate
sample = pd.DataFrame(
    [[55, 1, 4, 140]],
    columns=features
)

prediction = model.predict(sample)[0]

if prediction == 1:
    print("Prediction: Heart Disease")
else:
    print("Prediction: No Heart Disease")
```
